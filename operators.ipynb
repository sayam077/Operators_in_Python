```python

# Operators in Python:
 #1. Arithmetic Operators:
 # +	Addition = Adds two values; also joins strings together	
print(3 + 2)
 #-	Subtraction	= Subtracts right operand from the left	        
print(10 - 3)
# *	Multiplication = Multiplies; also repeats strings and lists	   
print(3 * 4)
 #/	Division = Divides and always returns a float	           
print(7 / 2)
# //	Floor Division = Divides and rounds DOWN to nearest integer	
print(7 // 2)
 # % Modulus = Returns the remainder after division	        
print(7 % 3)
 #** Exponentiation	= Raises left to the power of right	           
print(2 ** 10)

```

```python
# 2. Comparison (Relational) operators
#==	Equal to    	  
print(5 == 5)
#!=	Not equal to
print(5 != 3)
#>	Greater than   	
print(7 > 3)
#<	Less than	
print(3 < 7) 
#>=	Greater or equal
print(5 >= 5)
#<=	Less or equal
print(4 <= 5)
 
```

```python
# 3. Assignment Operators

# =	Assign	Store a value in a variable	
x = 5

# Compound assignment operators
score = 50
score += 10 # score is now 60
score *= 2 # score is now 120
score -= 20 # score is now 100
print(score) # → 100

# Multiple assignment — assign several variables in one line
a = b = c = 0 # all three become 0
print(a, b, c) # → 0 0 0

# Tuple unpacking — assign different values in one line
x, y, z = 1, 2, 3
print(x, y, z) # → 1 2 3

# Swap variables — Pythonic way, no temporary variable needed
a, b = 10, 20
a, b = b, a # swap!
print(a, b) # → 20 10
print(x)
```

```python
# 4. Logical Operators

# and — both conditions must be True
age = 20
has_id = True
print(age >= 18 and has_id) # → True (both conditions satisfied)

# or — at least one condition must be True
is_student = True
is_teacher = False
print(is_student or is_teacher) # → True (one is True)

# not — inverts the boolean value
print(not True) # → False
print(not False) # → True

# Short-circuit with 'and' — right side skipped if left is False
print(False and 1/0) # → False (no ZeroDivisionError!)

# Short-circuit with 'or' — right side skipped if left is True
print(True or 1/0) # → True (no ZeroDivisionError!)

# Combining logical operators with comparison operators
marks = 72
print(marks >= 40 and marks <= 100) # → True (valid score range)
print(marks < 40 or marks > 100) # → False (not out of range)
```

```python
# 5. Bitwise Operator:

# bin() shows the binary representation of any number
print(bin(5)) # → 0b101
print(bin(3)) # → 0b11

# AND — 1 only where both bits are 1
print(5 & 3) # → 1 (0101 & 0011 = 0001)

# OR — 1 where either bit is 1
print(5 | 3) # → 7 (0101 | 0011 = 0111)

# XOR — 1 where bits are different
print(5 ^ 3) # → 6 (0101 ^ 0011 = 0110)

# Left shift — multiply by powers of 2
print(5 << 1) # → 10 (5 × 2)
print(5 << 2) # → 20 (5 × 4)

# Right shift — divide by powers of 2
print(20 >> 1) # → 10 (20 ÷ 2)
print(20 >> 2) # → 5 (20 ÷ 4)
```

```python
# 6. Membership Operators:

# Membership operators test whether a value exists inside a sequence 
# — such as a string, list, or tuple.
#  They return True or False. 
# The two membership operators are in and not in.

# Membership — checking if a value exists in a sequence
print("py" in "python") # → True
print("Java" in "python") # → False
print("z" not in "hello") # → True
```

```python
# 7.Identity Operators:

# Identity operators test whether two variables point to the exact same object in memory 
# — not just whether they have equal values, but whether they are literally the same object.
#  The two identity operators are is and is not.

# Identity — == checks value equality, is checks object identity
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b) # → True (same values)
print(a is b) # → False (different objects in memory)
print(a is c) # → True (same object — c points to a)
# The correct way to check for None — always use 'is'
x = None
print(x is None) # → True ✅ correct
print(x == None) # → True ⚠️ works but not recommended
```
